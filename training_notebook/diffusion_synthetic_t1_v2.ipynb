# Synthetic 3D T1 Brain Tumor MRI (NIfTI Latent Diffusion)

This notebook trains an unconditional 3D latent diffusion model for synthetic T1 brain tumor MRI volumes directly from NIfTI data.

## Workflow (NIfTI 3D, Unconditional T1 Generation)
1. Discover 3D T1 `.nii`/`.nii.gz` files recursively.
2. Train a 3D `AutoencoderKL` on T1 volumes.
3. Freeze the autoencoder and train a latent DDPM UNet.
4. Use EMA UNet for stable sampling and decode synthetic T1 volumes.
5. Save generated volumes as NIfTI with optional reference affine/header.

```python
# =========================
# Config (mount MyDrive for outputs + gdown from shared link + auto extract + 70/20/10 split downstream)
# =========================
DATASET_ROOT_OVERRIDE = None  # set only if auto-discovery fails
DRIVE_FILE_LINK = "https://drive.google.com/file/d/1Ox7sCEqtM7gOM8Y72H1bqHULtWolfTu3/view?usp=sharing"  # shared source link

# Keep downloads/extraction on runtime disk for speed; keep outputs on your MyDrive.
DOWNLOAD_ROOT = Path("/content/dataset_download")  # temporary download location
EXTRACT_ROOT = Path("/content/extracted_datasets")  # temporary extraction location

AUTO_DOWNLOAD = True
AUTO_EXTRACT = True

# Drive behavior for outputs
MOUNT_MYDRIVE_FOR_OUTPUTS = True  # ensure outputs/checkpoints are saved to your drive
FORCE_REMOUNT_MYDRIVE = False  # set True if you need to re-select/switch account

import tarfile
import zipfile

# Detect Colab so Drive mounting and runtime paths can be configured automatically.
IS_COLAB = (
    "COLAB_GPU" in os.environ
    or "COLAB_RELEASE_TAG" in os.environ
    or Path.cwd().as_posix().startswith("/content")
)

if IS_COLAB and MOUNT_MYDRIVE_FOR_OUTPUTS:
    try:
        import importlib
        import subprocess
        import sys
        drive = importlib.import_module("google.colab.drive")
        need_mount = FORCE_REMOUNT_MYDRIVE or (not Path("/content/drive/MyDrive").exists())
        if need_mount:
            print("Mounting Google Drive at /content/drive ...")
            drive.mount("/content/drive", force_remount=FORCE_REMOUNT_MYDRIVE)
        else:
            print("Google Drive already mounted at /content/drive.")
    except Exception as e:
        print(f"Drive mount step skipped or failed: {e}")


def _find_dir_recursively(root: Path, dir_name: str):
    """
    Function description:
        Find a directory by name anywhere under a root path.
    
    Parameters:
        root (Path): Root directory to search.
        dir_name (str): Directory name to match case-insensitively.
    
    Returns:
        Path | None: Matching directory path when found, otherwise None.
    """
    if not root.exists() or not root.is_dir():
        return None
    for p in root.rglob("*"):
        if p.is_dir() and p.name.lower() == dir_name.lower():
            return p
    return None


def _find_any_nifti_parent(root: Path):
    """
    Function description:
        Locate the parent folder of any NIfTI file under a root path.
    
    Parameters:
        root (Path): Root directory to search.
    
    Returns:
        Path | None: Parent directory of the first discovered NIfTI file.
    """
    if not root.exists() or not root.is_dir():
        return None
    nii_files = list(root.rglob("*.nii")) + list(root.rglob("*.nii.gz"))
    if not nii_files:
        return None
    return nii_files[0].parent.resolve()


def _pick_downloaded_archive(download_root: Path):
    """
    Function description:
        Select the most recently modified supported archive from the download folder.
    
    Parameters:
        download_root (Path): Folder containing downloaded files.
    
    Returns:
        Path | None: Archive path when a supported archive exists, otherwise None.
    """
    candidates = []
    for p in download_root.rglob("*"):
        if not p.is_file():
            continue
        name = p.name.lower()
        if name.endswith(".zip") or name.endswith(".tar") or name.endswith(".tar.gz") or name.endswith(".tgz") or name.endswith(".tar.bz2") or name.endswith(".tbz2") or name.endswith(".tar.xz") or name.endswith(".txz"):
            candidates.append(p)

    if not candidates:
        return None

    candidates = sorted(candidates, key=lambda p: p.stat().st_mtime, reverse=True)
    return candidates[0]


def _extract_archive(archive_path: Path, extract_root: Path):
    """
    Function description:
        Extract a supported dataset archive into the configured extraction directory.
    
    Parameters:
        archive_path (Path): Archive file to extract.
        extract_root (Path): Destination directory for extracted files.
    
    Returns:
        None
    """
    if not AUTO_EXTRACT:
        raise FileNotFoundError(f"Auto extract disabled, but need extraction for: {archive_path}")

    extract_root.mkdir(parents=True, exist_ok=True)
    name = archive_path.name.lower()
    print(f"Extracting archive: {archive_path}")

    if name.endswith(".zip"):
        with zipfile.ZipFile(archive_path, "r") as zf:
            zf.extractall(extract_root)
        return

    if (
        name.endswith(".tar")
        or name.endswith(".tar.gz")
        or name.endswith(".tgz")
        or name.endswith(".tar.bz2")
        or name.endswith(".tbz2")
        or name.endswith(".tar.xz")
        or name.endswith(".txz")
    ):
        with tarfile.open(archive_path, "r:*") as tf:
            tf.extractall(extract_root)
        return

    raise ValueError(f"Unsupported archive format: {archive_path}")


def _discover_dataset_root() -> Path:
    """
    Function description:
        Resolve the dataset root from an override, downloaded archive, or extracted files.
    
    Parameters:
        None
    
    Returns:
        Path: Resolved dataset root containing T1 NIfTI volumes.
    """
    if DATASET_ROOT_OVERRIDE is not None:
        override = Path(DATASET_ROOT_OVERRIDE).expanduser()
        if not override.exists():
            raise FileNotFoundError(f"DATASET_ROOT_OVERRIDE does not exist: {override}")
        return override.resolve()

    DOWNLOAD_ROOT.mkdir(parents=True, exist_ok=True)
    EXTRACT_ROOT.mkdir(parents=True, exist_ok=True)

    if AUTO_DOWNLOAD:
        print("Installing gdown...")
        subprocess.run([sys.executable, "-m", "pip", "install", "-q", "gdown"], check=True)
        print("Downloading dataset archive with gdown from shared link...")
        print(f"Download target: {DOWNLOAD_ROOT} (runtime disk)")
        subprocess.run([sys.executable, "-m", "gdown", "--fuzzy", DRIVE_FILE_LINK, "-O", f"{DOWNLOAD_ROOT}/"], check=True)

    archive_path = _pick_downloaded_archive(DOWNLOAD_ROOT)
    if archive_path is not None:
        _extract_archive(archive_path, EXTRACT_ROOT)
    else:
        print("No archive file detected in download root; trying direct-folder/direct-file discovery.")

    # Prefer known dataset layouts before falling back to any folder containing NIfTI files.
    # 1) Prefer final-clean-data folder name if present
    final_clean = _find_dir_recursively(EXTRACT_ROOT, "final-clean-data")
    if final_clean is not None:
        return final_clean.resolve()

    # 2) Otherwise use folder that contains T1
    t1_dir = _find_dir_recursively(EXTRACT_ROOT, "T1")
    if t1_dir is not None and t1_dir.parent.exists():
        return t1_dir.parent.resolve()

    # 3) Fallback to direct NIfTI parent
    nii_parent = _find_any_nifti_parent(EXTRACT_ROOT)
    if nii_parent is not None:
        return nii_parent

    # 4) Final fallback: maybe data downloaded directly without extraction
    t1_dir = _find_dir_recursively(DOWNLOAD_ROOT, "T1")
    if t1_dir is not None and t1_dir.parent.exists():
        return t1_dir.parent.resolve()

    nii_parent = _find_any_nifti_parent(DOWNLOAD_ROOT)
    if nii_parent is not None:
        return nii_parent

    raise FileNotFoundError(
        "Could not find dataset root after gdown download/extract. "
        "Expected a folder containing T1 and/or .nii/.nii.gz files."
    )


DATASET_ROOT = _discover_dataset_root()
T1_ROOT = _find_dir_recursively(DATASET_ROOT, "T1")

# If no explicit T1 directory exists, but DATASET_ROOT itself contains nii files, use it directly.
if T1_ROOT is None:
    has_nii_here = any(DATASET_ROOT.glob("*.nii")) or any(DATASET_ROOT.glob("*.nii.gz"))
    if has_nii_here:
        T1_ROOT = DATASET_ROOT
    else:
        raise FileNotFoundError(f"Could not locate T1 directory or NIfTI files under dataset root: {DATASET_ROOT}")

DRIVE_OUTPUT_ROOT = Path("/content/drive/MyDrive/MDS03_Backend_outputs")  # your drive output root
OUTPUT_ROOT = DRIVE_OUTPUT_ROOT / "synthetic_t1_3d_ldm"  # this notebook's output folder
OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)

# Reproducibility
# Fix randomness across Python, NumPy, MONAI, and Torch for reproducible runs.
# Reproducibility
SEED = 42
set_determinism(seed=SEED)
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

# Data
T1_FILENAME_FILTER = "t1"   # use None to keep all nii files

# Split ratio
TRAIN_RATIO = 0.7
VAL_RATIO = 0.2
TEST_RATIO = 0.1

# Spatial setup (H, W, D)
# User-updated canonical names and aliases
LATENT_CHANNELS = 8
AUTOENCODER_CHANNELS = (32, 64, 128)
DIFFUSION_IN_CHANNELS = 8
DIFFUSION_OUT_CHANNELS = 8
SPATIAL_SIZE = (96, 96, 64)
PIXDIM = (1.0, 1.0, 1.0)
NUM_TRAIN_TIMESTEPS = 1000
BETA_START = 0.0012
BETA_END = 0.0185

# Training (user-specified values)
AE_EPOCHS = 160
DIFF_EPOCHS = 260
AE_LR = 5e-5
DIFF_LR = 2e-5
BATCH_SIZE = 1
NUM_WORKERS = 2
SEED = 42
KL_WEIGHT = 7.5e-7
NUM_INFERENCE_STEPS = 1000
USE_EMA = True
EMA_DECAY = 0.9995
OFFSET_NOISE = 0.05
GRAD_CLIP = 1.0
SAVE_EVERY = 5
AE_EARLY_STOP_PATIENCE = 30
AE_EARLY_STOP_MIN_DELTA = 2e-5
DIFF_EARLY_STOP_PATIENCE = 60
DIFF_EARLY_STOP_MIN_DELTA = 5e-5

GRAD_ACCUM_STEPS = 2
TARGET_SHAPE = SPATIAL_SIZE
TARGET_PIXDIM = PIXDIM
NUM_DIFFUSION_TIMESTEPS = NUM_TRAIN_TIMESTEPS
INFERENCE_STEPS = 1000
# Intensity normalization percentiles
NORM_LOW_PCT = 0.5
NORM_HIGH_PCT = 99.5

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
use_amp = device.type == "cuda"
print("Using device:", device)
print("AMP enabled:", use_amp)
print("Working dir:", Path.cwd())
print("Running in Colab-like runtime:", IS_COLAB)
print("Dataset root:", DATASET_ROOT)
print("T1 root:", T1_ROOT)
print("Output root:", OUTPUT_ROOT)
print(f"Split ratio train/val/test: {TRAIN_RATIO:.1f}/{VAL_RATIO:.1f}/{TEST_RATIO:.1f}")
print("Drive mount for outputs:", MOUNT_MYDRIVE_FOR_OUTPUTS)
print("Force remount:", FORCE_REMOUNT_MYDRIVE)
print(f"Dataset link source: {DRIVE_FILE_LINK}")
```

```python
# Data discovery + split
def natural_key(text: str):
    """
    Function description:
        Split text into numeric and non-numeric parts for natural sorting.
    
    Parameters:
        text (str): Text value to split.
    
    Returns:
        list[int | str]: Sort key that keeps numbered filenames in human order.
    """
    return [int(c) if c.isdigit() else c for c in re.split(r"(\\d+)", text)]


def discover_nifti_files(dataset_root: Path, filename_filter: str | None = "t1") -> list[Path]:
    """
    Function description:
        Discover NIfTI files under a dataset root with an optional filename filter.
    
    Parameters:
        dataset_root (Path): Root folder to search recursively.
        filename_filter (str | None): Optional substring required in each filename.
    
    Returns:
        list[Path]: Sorted NIfTI file paths.
    """
    nii_files = sorted(dataset_root.rglob("*.nii*"), key=lambda p: natural_key(str(p).lower()))
    if filename_filter:
        k = filename_filter.lower()
        nii_files = [p for p in nii_files if k in p.name.lower()]

    if not nii_files:
        raise ValueError(
            f"No NIfTI files found under {dataset_root} with filter={filename_filter!r}"
        )
    return nii_files


def build_data_dicts(paths: list[Path]) -> list[dict]:
    """
    Function description:
        Convert NIfTI paths into MONAI-style data dictionaries.
    
    Parameters:
        paths (list[Path]): NIfTI file paths to convert.
    
    Returns:
        list[dict]: Records containing image paths and case identifiers.
    """
    return [{"image": str(p), "case_id": p.stem} for p in paths]


if T1_ROOT is None:
    raise FileNotFoundError("T1_ROOT is None. Ensure dataset discovery cell has run successfully.")
t1_root: Path = T1_ROOT

all_t1_nii = discover_nifti_files(t1_root, filename_filter=T1_FILENAME_FILTER)
# Build MONAI dictionaries after discovery so transforms can load image paths directly.
all_files = build_data_dicts(all_t1_nii)

# Deterministic shuffle before split
rng = random.Random(SEED)
rng.shuffle(all_files)

n_total = len(all_files)
n_train = int(n_total * TRAIN_RATIO)
n_val = int(n_total * VAL_RATIO)
n_test = n_total - n_train - n_val

# Validate split sizes early to avoid silent empty train/validation/test loaders.
if n_total < 10:
    raise ValueError(f"Too few samples ({n_total}) to create reliable 70/20/10 split.")
if n_train == 0 or n_val == 0 or n_test == 0:
    raise ValueError(
        f"Invalid split sizes from {n_total} samples -> train={n_train}, val={n_val}, test={n_test}"
    )

train_files = all_files[:n_train]
val_files = all_files[n_train:n_train + n_val]
test_files = all_files[n_train + n_val:]

train_nii = [Path(d["image"]) for d in train_files]
val_nii = [Path(d["image"]) for d in val_files]
test_nii = [Path(d["image"]) for d in test_files]

# Keep CPU fallback practical: train on a tiny subset so notebook does real work without taking hours.
if "device" in globals() and device.type != "cuda" and globals().get("CPU_FAST_MODE", False):
    max_train_cases = min(8, len(train_files))
    max_val_cases = min(4, len(val_files))
    max_test_cases = min(4, len(test_files))
    train_files = train_files[:max_train_cases]
    val_files = val_files[:max_val_cases]
    test_files = test_files[:max_test_cases]
    train_nii = train_nii[:max_train_cases]
    val_nii = val_nii[:max_val_cases]
    test_nii = test_nii[:max_test_cases]
    print(f"CPU fast-mode subset active: train={len(train_files)}, val={len(val_files)}, test={len(test_files)}")

print(f"Total T1 NIfTI volumes: {n_total}")
print(f"Train volumes (70%): {len(train_files)}")
print(f"Val volumes (20%)  : {len(val_files)}")
print(f"Test volumes (10%) : {len(test_files)}")

print("Train preview:")
for p in train_nii[:3]:
    print(" -", p)
print("Validation preview:")
for p in val_nii[:3]:
    print(" -", p)
print("Test preview:")
for p in test_nii[:3]:
    print(" -", p)
```

```python
# Transforms + dataloaders
train_transforms = Compose([
    LoadImaged(keys=["image"]),
    EnsureChannelFirstd(keys=["image"]),
    EnsureTyped(keys=["image"]),
    Orientationd(keys=["image"], axcodes="RAS"),
    Spacingd(keys=["image"], pixdim=TARGET_PIXDIM, mode=("bilinear",)),
    Resized(keys=["image"], spatial_size=TARGET_SHAPE, mode=("trilinear",), anti_aliasing=True),
    ScaleIntensityRangePercentilesd(
        keys=["image"],
        lower=NORM_LOW_PCT,
        upper=NORM_HIGH_PCT,
        b_min=-1.0,
        b_max=1.0,
        clip=True,
    ),
    RandFlipd(keys=["image"], prob=0.5, spatial_axis=0),
    RandFlipd(keys=["image"], prob=0.5, spatial_axis=1),
    RandFlipd(keys=["image"], prob=0.5, spatial_axis=2),
    RandRotate90d(keys=["image"], prob=0.3, max_k=3),
    RandAffined(
        keys=["image"],
        prob=0.15,
        rotate_range=(0.05, 0.05, 0.05),
        scale_range=(0.05, 0.05, 0.05),
        mode=("bilinear",),
        padding_mode="border",
    ),
    EnsureTyped(keys=["image"]),
])

val_transforms = Compose([
    LoadImaged(keys=["image"]),
    EnsureChannelFirstd(keys=["image"]),
    EnsureTyped(keys=["image"]),
    Orientationd(keys=["image"], axcodes="RAS"),
    Spacingd(keys=["image"], pixdim=TARGET_PIXDIM, mode=("bilinear",)),
    Resized(keys=["image"], spatial_size=TARGET_SHAPE, mode=("trilinear",), anti_aliasing=True),
    ScaleIntensityRangePercentilesd(
        keys=["image"],
        lower=NORM_LOW_PCT,
        upper=NORM_HIGH_PCT,
        b_min=-1.0,
        b_max=1.0,
        clip=True,
    ),
    EnsureTyped(keys=["image"]),
])

train_ds = Dataset(data=train_files, transform=train_transforms)
val_ds = Dataset(data=val_files, transform=val_transforms)

train_loader = DataLoader(
    train_ds,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=NUM_WORKERS,
    pin_memory=False,
    persistent_workers=(NUM_WORKERS > 0),
)
val_loader = DataLoader(
    val_ds,
    batch_size=1,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=False,
    persistent_workers=(NUM_WORKERS > 0),
)

print("Train batches:", len(train_loader))
print("Val batches  :", len(val_loader))

batch = next(iter(train_loader))
x = batch["image"]
print("Sample tensor shape:", tuple(x.shape))  # [B, 1, H, W, D]
print("Intensity min/max:", float(x.min()), float(x.max()))
if float(x.min()) < -1.05 or float(x.max()) > 1.05:
    print("Warning: intensity appears outside expected [-1, 1] range.")
else:
    print("Intensity check passed: data is in expected [-1, 1] range.")
```

```python
# =========================
# 3D Autoencoder (T1)
# =========================

# Clear stale model/optimizer tensors when re-running notebook cells repeatedly.
for _name in ["autoencoder", "unet", "ema_unet", "optimizer_ae", "optimizer_diff", "scaler_ae", "scaler_diff"]:
    if _name in globals():
        try:
            obj = globals()[_name]
            if hasattr(obj, "to"):
                obj.to("cpu")
        except Exception:
            pass
        try:
            del globals()[_name]
        except Exception:
            pass
gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()

autoencoder = AutoencoderKL(
    spatial_dims=3,
    in_channels=1,
    out_channels=1,
    num_channels=AE_NUM_CHANNELS,
    latent_channels=AE_LATENT_CHANNELS,
    num_res_blocks=2,
    norm_num_groups=16,
    attention_levels=(False, False, True),
).to(device)

optimizer_ae = torch.optim.AdamW(autoencoder.parameters(), lr=AE_LR, weight_decay=1e-6)
scaler_ae = GradScaler('cuda', enabled=use_amp)
kl_weight = KL_WEIGHT

# Decay AE LR to a smaller floor late in training so slow-improving runs can refine without drifting.
lr_scheduler_ae = torch.optim.lr_scheduler.MultiStepLR(
    optimizer_ae,
    milestones=[90],
    gamma=0.2,
    last_epoch=-1,
)

def kl_loss(z_mu, z_sigma):
    """
    Function description:
        Compute the KL-divergence loss term for the variational autoencoder latents.
    
    Parameters:
        z_mu (torch.Tensor): Latent mean tensor predicted by the autoencoder.
        z_sigma (torch.Tensor): Latent standard-deviation tensor predicted by the autoencoder.
    
    Returns:
        torch.Tensor: Mean KL loss for the batch.
    """
    eps = 1e-8
    kl = 0.5 * torch.sum(
        z_mu.pow(2) + z_sigma.pow(2) - torch.log(z_sigma.pow(2) + eps) - 1,
        dim=[1, 2, 3, 4],
    )
    return kl.mean()


def _to_01(x_np: np.ndarray) -> np.ndarray:
    """
    Function description:
        Convert model-range values from [-1, 1] to display range [0, 1].
    
    Parameters:
        x_np (np.ndarray): Input array in model output range.
    
    Returns:
        np.ndarray: Array clipped and remapped to [0, 1].
    """
    x_np = np.clip(x_np, -1.0, 1.0)
    return (x_np + 1.0) / 2.0


def _volume_psnr_ssim(gt_vol: np.ndarray, pred_vol: np.ndarray):
    """
    Function description:
        Compute PSNR and mean slice-wise SSIM between two 3D volumes.
    
    Parameters:
        gt_vol (np.ndarray): Ground-truth volume in model range [-1, 1].
        pred_vol (np.ndarray): Predicted volume in model range [-1, 1].
    
    Returns:
        tuple[float, float]: PSNR and mean SSIM values.
    """
    gt = _to_01(gt_vol)
    pred = _to_01(pred_vol)

    psnr_val = psnr(gt, pred, data_range=1.0)

    ssim_slices = []
    for z in range(gt.shape[2]):
        ssim_val = ssim(gt[:, :, z], pred[:, :, z], data_range=1.0)
        ssim_slices.append(ssim_val)
    return float(psnr_val), float(np.mean(ssim_slices))


def train_autoencoder_epoch(model, loader, optimizer, scaler, device):
    """
    Function description:
        Train the 3D autoencoder for one epoch on T1 volumes.
    
    Parameters:
        model (torch.nn.Module): Autoencoder model to train.
        loader (DataLoader): Training data loader.
        optimizer (torch.optim.Optimizer): Optimizer used for parameter updates.
        scaler (GradScaler): Mixed-precision gradient scaler.
        device (torch.device): Device used for tensors and model execution.
    
    Returns:
        float: Average autoencoder training loss.
    """
    model.train()
    for p in model.parameters():
        p.requires_grad = True
    running = 0.0
    
    total_batches = len(loader)
    print(f"🚀 Starting training... Total batches to process: {total_batches}")

    for step, batch in enumerate(loader): # <-- Removed tqdm completely
        t1 = batch["image"].to(device)
        optimizer.zero_grad(set_to_none=True)

        try:
            with autocast('cuda', enabled=use_amp, dtype=AMP_DTYPE):
                recon, z_mu, z_sigma = model(t1)
                
                # Train with reconstruction plus KL loss so the latent space remains usable for diffusion.
                recon_l = F.l1_loss(recon.float(), t1.float())
                k_loss = kl_loss(z_mu, z_sigma)
                
                loss = recon_l + (kl_weight * k_loss)

            scaler.scale(loss).backward()
            scaler.step(optimizer)
            scaler.update()
        except torch.cuda.OutOfMemoryError:
            if torch.cuda.is_available():
                torch.cuda.empty_cache()
            raise

        running += loss.item()
    
    return running / max(1, total_batches)

# AE validation helpers (defines PSNR/SSIM calculation used during training)
@torch.no_grad()
def validate_autoencoder_epoch(model, loader, device):
    """
    Function description:
        Validate the 3D autoencoder and compute reconstruction quality metrics.
    
    Parameters:
        model (torch.nn.Module): Autoencoder model to evaluate.
        loader (DataLoader): Validation data loader.
        device (torch.device): Device used for tensors and model execution.
    
    Returns:
        dict: Validation loss, PSNR, and SSIM values.
    """
    model.eval()
    running = 0.0
    psnr_vals = []
    ssim_vals = []
    pbar = tqdm(loader, desc="AE Val", ncols=110)

    for step, batch in enumerate(pbar):
        t1 = batch["image"].to(device)
        with autocast('cuda', enabled=use_amp, dtype=AMP_DTYPE):
            recon, z_mu, z_sigma = model(t1)
            
            recon_l = F.l1_loss(recon.float(), t1.float())
            k_loss = kl_loss(z_mu, z_sigma)
            
            loss = recon_l + (kl_weight * k_loss)

        running += loss.item()

        # Track reconstruction metrics on the validation set alongside the optimization loss.
        gt_np = t1[0, 0].detach().cpu().float().numpy()
        pred_np = recon[0, 0].detach().cpu().float().numpy()
        p_val, s_val = _volume_psnr_ssim(gt_np, pred_np)
        psnr_vals.append(p_val)
        ssim_vals.append(s_val)

        pbar.set_postfix(
            loss=f"{running / (step + 1):.4f}",
            psnr=f"{np.mean(psnr_vals):.2f}",
            ssim=f"{np.mean(ssim_vals):.4f}",
        )

    return {
        "loss": running / max(1, len(loader)),
        "psnr": float(np.mean(psnr_vals)) if psnr_vals else 0.0,
        "ssim": float(np.mean(ssim_vals)) if ssim_vals else 0.0,
    }
```

```python
def train_diffusion_epoch(
    autoencoder, unet, ema_unet, noise_scheduler, loader, optimizer, scaler, device,
    scale_factor, grad_accum_steps=1, ema_decay=0.999, offset_noise=0.1
):
    """
    Function description:
        Train the latent diffusion UNet for one epoch on encoded T1 latents.
    
    Parameters:
        autoencoder (torch.nn.Module): Frozen autoencoder used to encode T1 volumes.
        unet (torch.nn.Module): Diffusion UNet being trained.
        ema_unet (torch.nn.Module): EMA copy of the diffusion UNet.
        noise_scheduler: Scheduler used to add noise during training.
        loader (DataLoader): Training data loader.
        optimizer (torch.optim.Optimizer): Optimizer for UNet parameters.
        scaler (GradScaler): Mixed-precision gradient scaler.
        device (torch.device): Device used for tensors and models.
        scale_factor (float): Scale applied to encoded latents.
        grad_accum_steps (int): Number of batches accumulated before an optimizer step.
        ema_decay (float): Decay factor used for EMA updates.
        offset_noise (float): Strength of optional offset noise.
    
    Returns:
        float: Average training MSE loss for the epoch.
    """
    # Freeze the autoencoder; diffusion training updates only the latent denoiser.
    autoencoder.eval()
    for p in autoencoder.parameters():
        p.requires_grad = False

    unet.train()
    running = 0.0
    pbar = tqdm(loader, desc="Diff Train", ncols=120)
    optimizer.zero_grad(set_to_none=True)

    for step, batch in enumerate(pbar):
        t1 = batch["image"].to(device)

        with torch.no_grad():
            z_t1 = autoencoder.encode_stage_2_inputs(t1) * scale_factor

        noise = torch.randn_like(z_t1)
        if offset_noise > 0:
            noise = noise + offset_noise * torch.randn(
                z_t1.shape[0], z_t1.shape[1], 1, 1, 1, device=device, dtype=z_t1.dtype
            )

        timesteps = torch.randint(
            0, noise_scheduler.num_train_timesteps, (z_t1.shape[0],), device=device, dtype=torch.long
        )

        noisy = noise_scheduler.add_noise(
            original_samples=z_t1, noise=noise, timesteps=timesteps
        )

        with autocast('cuda', enabled=use_amp):
            noise_pred = unet(noisy, timesteps=timesteps)
            if hasattr(noise_pred, "sample"):
                noise_pred = noise_pred.sample
            loss = F.mse_loss(noise_pred.float(), noise.float()) / grad_accum_steps

        scaler.scale(loss).backward()

        # Accumulate gradients to simulate a larger batch under 3D memory limits.
        should_step = ((step + 1) % grad_accum_steps == 0) or (step + 1 == len(loader))
        if should_step:
            scaler.unscale_(optimizer)
            torch.nn.utils.clip_grad_norm_(unet.parameters(), max_norm=GRAD_CLIP)
            scaler.step(optimizer)
            scaler.update()
            optimizer.zero_grad(set_to_none=True)
            update_ema(ema_unet, unet, decay=ema_decay)

        running += loss.item() * grad_accum_steps
        avg_loss = running / (step + 1)
        pbar.set_postfix(train_loss=f"{avg_loss:.4f}", lr=f"{optimizer.param_groups[0]['lr']:.2e}")

    return running / max(1, len(loader))


@torch.no_grad()
def validate_diffusion_epoch(
    autoencoder, eval_unet, noise_scheduler, loader, device, scale_factor, seed=1234
):
    """
    Function description:
        Evaluate the latent diffusion UNet on validation batches.
    
    Parameters:
        autoencoder (torch.nn.Module): Frozen autoencoder used to encode T1 volumes.
        eval_unet (torch.nn.Module): UNet or EMA UNet used for validation.
        noise_scheduler: Scheduler used to add validation noise.
        loader (DataLoader): Validation data loader.
        device (torch.device): Device used for tensors and models.
        scale_factor (float): Scale applied to encoded latents.
        seed (int): Seed used for deterministic validation noise.
    
    Returns:
        float: Average validation MSE loss.
    """
    autoencoder.eval()
    eval_unet.eval()
    running = 0.0
    # Use deterministic validation noise so loss comparisons across epochs are stable.
    g = torch.Generator(device=device).manual_seed(seed)

    pbar = tqdm(loader, desc="Diff Val", ncols=120)
    for step, batch in enumerate(pbar):
        t1 = batch["image"].to(device)

        z_t1 = autoencoder.encode_stage_2_inputs(t1) * scale_factor
        noise = torch.randn(z_t1.shape, generator=g, device=device, dtype=z_t1.dtype)
        timesteps = torch.randint(
            0, noise_scheduler.num_train_timesteps, (z_t1.shape[0],), generator=g, device=device
        ).long()

        noisy = noise_scheduler.add_noise(
            original_samples=z_t1, noise=noise, timesteps=timesteps
        )

        with autocast('cuda', enabled=use_amp):
            noise_pred = eval_unet(noisy, timesteps=timesteps)
            if hasattr(noise_pred, "sample"):
                noise_pred = noise_pred.sample
            loss = F.mse_loss(noise_pred.float(), noise.float())

        running += loss.item()
        pbar.set_postfix(val_loss=f"{running / (step + 1):.4f}")

    return running / max(1, len(loader))


@torch.no_grad()
def estimate_scale_factor(autoencoder, loader, device, num_samples=256):
    """
    Function description:
        Estimate the latent scale factor from encoded training samples.
    
    Parameters:
        autoencoder (torch.nn.Module): Autoencoder used to encode training volumes.
        loader (DataLoader): Data loader that yields T1 batches.
        device (torch.device): Device used for encoding.
        num_samples (int): Maximum number of samples to include in the estimate.
    
    Returns:
        float: Mean absolute latent magnitude used as the scale factor.
    """
    autoencoder.eval()
    latent_list = []
    
    for step, batch in enumerate(loader):
        if step * loader.batch_size >= num_samples:
            break
        t1 = batch["image"].to(device)
        z = autoencoder.encode_stage_2_inputs(t1)
        latent_list.append(z.detach().cpu().abs().mean())
    
    if not latent_list:
        print("Warning: no samples processed for scale factor estimation. Using default=1.0")
        return 1.0
    
    scale_factor = float(torch.mean(torch.stack(latent_list)))
    return scale_factor


# Helper: EMA update
def update_ema(ema_model, model, decay=0.9999):
    """
    Function description:
        Update an exponential moving average model from the current model weights.
    
    Parameters:
        ema_model (torch.nn.Module): EMA model to update in place.
        model (torch.nn.Module): Current model whose weights are tracked.
        decay (float): EMA decay factor.
    
    Returns:
        None
    """
    with torch.no_grad():
        for ema_param, param in zip(ema_model.parameters(), model.parameters()):
            ema_param.data.mul_(decay).add_(param.data, alpha=1 - decay)


# Helper: Resize volume to target shape
def _resize_volume_to_shape(src_volume: np.ndarray, target_shape: tuple[int, ...]) -> np.ndarray:
    """
    Function description:
        Resize a 3D volume to the requested target shape.
    
    Parameters:
        src_volume (np.ndarray): Source 3D volume.
        target_shape (tuple[int, ...]): Desired output shape.
    
    Returns:
        np.ndarray: Resized volume.
    """
    if tuple(src_volume.shape) == tuple(target_shape):
        return src_volume
    src_t = torch.from_numpy(src_volume).unsqueeze(0).unsqueeze(0).float()
    out = F.interpolate(src_t, size=target_shape, mode="trilinear", align_corners=False)
    return out.squeeze(0).squeeze(0).cpu().numpy()


# Helper: Generate synthetic T1 samples via diffusion
@torch.no_grad()
def generate_t1(
    autoencoder,
    unet,
    scheduler,
    scale_factor,
    device,
    latent_spatial_shape,
    batch_size=1,
    latent_channels=8,
    num_inference_steps=300,
    sample_seed: int | None = None,
):
    """
    Function description:
        Generate synthetic T1 volumes by denoising random latent tensors and decoding them.
    
    Parameters:
        autoencoder (torch.nn.Module): Autoencoder used to decode generated latents.
        unet (torch.nn.Module): Diffusion UNet used for reverse diffusion.
        scheduler: Scheduler with timesteps configured for inference.
        scale_factor (float): Scale applied before decoding latents.
        device (torch.device): Device used for generation.
        latent_spatial_shape (tuple[int, int, int]): Spatial shape of sampled latent noise.
        batch_size (int): Number of volumes to generate together.
        latent_channels (int): Number of channels in the latent tensor.
        num_inference_steps (int): Number of reverse-diffusion steps.
        sample_seed (int | None): Optional seed for deterministic generation.
    
    Returns:
        torch.Tensor: Generated T1 volumes clamped to [-1, 1].
    """
    autoencoder.eval()
    unet.eval()

    generator = None
    if sample_seed is not None:
        generator = torch.Generator(device=device).manual_seed(int(sample_seed))

    latent = torch.randn(
        (batch_size, latent_channels, *latent_spatial_shape),
        device=device,
        generator=generator,
    )
    scheduler.set_timesteps(num_inference_steps=num_inference_steps)

    for t in scheduler.timesteps:
        t_int = int(t.item()) if torch.is_tensor(t) else int(t)
        timesteps = torch.full((latent.shape[0],), t_int, device=device, dtype=torch.long)

        with autocast('cuda', enabled=use_amp):
            noise_pred = unet(latent, timesteps=timesteps)
            if hasattr(noise_pred, "sample"):
                noise_pred = noise_pred.sample

        step_out = scheduler.step(noise_pred, t_int, latent)
        latent = step_out[0] if isinstance(step_out, tuple) else step_out.prev_sample

    # Decode the final latent back to image space after undoing latent scaling.
    latent = latent / scale_factor
    pred_t1 = autoencoder.decode_stage_2_outputs(latent)
    pred_t1 = torch.clamp(pred_t1, -1.0, 1.0)
    return pred_t1
```

```python
def save_nifti_volume(
    volume_tensor,
    save_path: Path,
    reference_nifti_path: str | None = None,
    match_reference_shape: bool = True,
):
    """
    Function description:
        Save a generated tensor volume as a NIfTI file, optionally matching a reference scan.
    
    Parameters:
        volume_tensor (torch.Tensor): Generated volume tensor to save.
        save_path (Path): Destination NIfTI path.
        reference_nifti_path (str | None): Optional reference NIfTI used for affine/header/shape.
        match_reference_shape (bool): Whether to resize the generated volume to reference shape.
    
    Returns:
        None
    """
    vol = volume_tensor.detach().cpu().float().numpy()
    if vol.ndim == 4:
        vol = vol[0]

    original_shape = tuple(int(s) for s in vol.shape)

    if reference_nifti_path is not None and Path(reference_nifti_path).exists():
        from typing import cast
        from nibabel.spatialimages import SpatialImage

        ref = cast(SpatialImage, nib.load(reference_nifti_path))
        # Get target shape from the NIfTI header (avoids accessing runtime-only attributes)
        target_shape = tuple(int(s) for s in ref.shape[:3])

        # Matching the reference shape keeps saved outputs compatible with original scan geometry.
        if match_reference_shape and original_shape != target_shape:
            vol = _resize_volume_to_shape(vol, target_shape)
            print(f"Resized generated volume from {original_shape} to {target_shape} to match reference.")

        vol = vol.astype(np.float32)
        nii = nib.Nifti1Image(vol, affine=ref.affine, header=ref.header.copy())
    else:
        vol = vol.astype(np.float32)
        nii = nib.Nifti1Image(vol, affine=np.eye(4))

    save_path.parent.mkdir(parents=True, exist_ok=True)
    nib.save(nii, str(save_path))
```

```python
# =========================
# Generate synthetic volume matched to target image size + masked metrics
# =========================
import numpy as np
import nibabel as nib
from pathlib import Path

if len(val_files) == 0:
    raise RuntimeError("val_files is empty. Run data preparation cells first.")
if not ("autoencoder" in globals() and "ema_unet" in globals() and "noise_scheduler" in globals()):
    raise RuntimeError("Model objects not found. Run training/sampling cell first.")

from typing import cast
from nibabel.spatialimages import SpatialImage

def _load_nifti_volume_local(path: Path) -> np.ndarray:
    """
    Function description:
        Load a local NIfTI file as a 3D float32 volume.
    
    Parameters:
        path (Path): NIfTI file path to load.
    
    Returns:
        np.ndarray: Loaded 3D volume.
    """
    img = cast(SpatialImage, nib.load(str(path)))
    vol = img.get_fdata().astype(np.float32)
    if vol.ndim != 3:
        raise ValueError(f"Expected 3D volume, got shape {vol.shape} for {path}")
    return vol

def _normalize_01_local(vol: np.ndarray) -> np.ndarray:
    """
    Function description:
        Normalize a local volume to [0, 1] with percentile clipping.
    
    Parameters:
        vol (np.ndarray): Input volume to normalize.
    
    Returns:
        np.ndarray: Normalized volume in [0, 1].
    """
    vmin = np.percentile(vol, 0.5)
    vmax = np.percentile(vol, 99.5)
    vol = np.clip(vol, vmin, vmax)
    return (vol - vol.min()) / (vol.max() - vol.min() + 1e-8)

target_idx = 0
target_path = Path(val_files[target_idx]["image"])
out_path = OUTPUT_ROOT / "generated" / f"sample_synth_t1_match_case_{target_idx:03d}.nii.gz"

# Latent spatial shape is derived from the transformed target volume.
# Derive latent shape from the transformed target so generated volume size matches the model path.# Latent spatial shape is derived from the transformed target volume.
target_sample = val_transforms({"image": str(target_path)})
target_img = target_sample["image"] if isinstance(target_sample, dict) else target_sample
target_img = torch.as_tensor(target_img) if not torch.is_tensor(target_img) else target_img
target_tensor = target_img.unsqueeze(0).to(device)
with torch.no_grad():
    latent_shape = autoencoder.encode_stage_2_inputs(target_tensor).shape[2:]

pred_t1 = generate_t1(
    autoencoder=autoencoder,
    unet=ema_unet,
    scheduler=noise_scheduler,
    scale_factor=float(globals().get("scale_factor_runtime", scale_factor)),
    device=device,
    latent_spatial_shape=latent_shape,
    batch_size=1,
    latent_channels=latent_channels,
    num_inference_steps=INFERENCE_STEPS,
)

save_nifti_volume(
    pred_t1[0],
    out_path,
    reference_nifti_path=str(target_path),
    match_reference_shape=True,
)

real_vol = _load_nifti_volume_local(target_path)
synth_vol = _load_nifti_volume_local(out_path)

if synth_vol.shape != real_vol.shape:
    raise RuntimeError(
        f"Shape mismatch remains after save: synth={synth_vol.shape}, target={real_vol.shape}"
    )

real_n = _normalize_01_local(real_vol)
synth_n = _normalize_01_local(synth_vol)
# Masked metrics focus on brain tissue rather than background voxels.
brain_mask = real_n > 0.05

masked_mae = float(np.mean(np.abs(real_n[brain_mask] - synth_n[brain_mask])))
masked_mse = float(np.mean((real_n[brain_mask] - synth_n[brain_mask]) ** 2))
masked_psnr = float(10.0 * np.log10(1.0 / (masked_mse + 1e-8)))

print("Target path          :", target_path)
print("Saved synthetic path :", out_path)
print("Target shape         :", real_vol.shape)
print("Saved synthetic shape:", synth_vol.shape)
print("Masked MAE (brain)   :", masked_mae)
print("Masked PSNR (brain)  :", masked_psnr)
print("Note: this is still an unconditional model, so pairwise similarity is a strict metric.")
```

```python
# =========================
# FID vs 10% test split (generate 1000 synthetic samples)
# =========================
import sys
import subprocess
import numpy as np
import torch
import torch.nn.functional as F
from pathlib import Path
from tqdm.auto import tqdm

# Install FID dependencies in-kernel if missing.
try:
    from importlib import import_module
    FrechetInceptionDistance = import_module("torchmetrics.image.fid").FrechetInceptionDistance
except Exception:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "torchmetrics", "torch-fidelity"])
    from torchmetrics.image.fid import FrechetInceptionDistance

# ------------ User controls ------------
FID_NUM_SYNTH = 200          # requested number of generated samples
FID_NUM_REAL = 200               # number of real test samples used for FID
FID_SLICE_AXIS = 2                # 0=sagittal, 1=coronal, 2=axial
FID_IMAGE_SIZE = 299              # Inception input size
FID_GEN_BATCH = 2                 # keep small for 3D memory safety
FID_RANDOM_SEED = 123
SAVE_SYNTHETIC_NIFTI = False      # set True if you want to save all 1000 generated volumes
SYNTH_FID_DIR = OUTPUT_ROOT / "generated" / "fid_1000_samples"
SYNTH_FID_DIR.mkdir(parents=True, exist_ok=True)

if len(test_files) == 0:
    raise RuntimeError("test_files is empty. Ensure the 70/20/10 split cell has run successfully.")
if device.type != "cuda":
    raise RuntimeError("FID run is configured for CUDA. Please switch to a GPU runtime for a full 1000-sample evaluation.")
if not ("autoencoder" in globals() and "ema_unet" in globals() and "noise_scheduler" in globals()):
    raise RuntimeError("Model objects not found. Run the model-definition and training cells first.")

rng = np.random.default_rng(FID_RANDOM_SEED)

def _to_01_np(vol: np.ndarray) -> np.ndarray:
    """
    Function description:
        Convert model-range NumPy volume values from [-1, 1] to [0, 1].
    
    Parameters:
        vol (np.ndarray): Input volume in model range.
    
    Returns:
        np.ndarray: Volume clipped and remapped to [0, 1].
    """
    vol = np.clip(vol, -1.0, 1.0)
    return (vol + 1.0) / 2.0

def _extract_slice(vol_01: np.ndarray, axis: int, idx: int) -> np.ndarray:
    """
    Function description:
        Extract one 2D slice from a normalized 3D volume.
    
    Parameters:
        vol_01 (np.ndarray): Normalized 3D volume in [0, 1].
        axis (int): Axis along which to extract the slice.
        idx (int): Slice index along the selected axis.
    
    Returns:
        np.ndarray: Extracted 2D slice as float32.
    """
    if axis == 0:
        sl = vol_01[idx, :, :]
    elif axis == 1:
        sl = vol_01[:, idx, :]
    else:
        sl = vol_01[:, :, idx]
    return sl.astype(np.float32)

def _slice_to_fid_tensor_uint8(slice_2d: np.ndarray, image_size: int = 299) -> torch.Tensor:
    """
    Function description:
        Convert a 2D slice into a uint8 RGB tensor for TorchMetrics FID.
    
    Parameters:
        slice_2d (np.ndarray): Normalized 2D slice in [0, 1].
        image_size (int): Target square image size expected by the FID model.
    
    Returns:
        torch.Tensor: Uint8 RGB tensor with shape [1, 3, image_size, image_size].
    """
    t = torch.from_numpy(slice_2d).unsqueeze(0).unsqueeze(0).float()  # [1,1,H,W]
    t = F.interpolate(t, size=(image_size, image_size), mode="bilinear", align_corners=False)
    t = t.repeat(1, 3, 1, 1)  # [1,3,H,W]
    t = (t * 255.0).clamp(0, 255).to(torch.uint8)
    return t

# Cache transformed real volumes once so FID updates do not repeat preprocessing work.
print("Building transformed test cache from the 10% test split...")
test_ds_fid = Dataset(data=test_files, transform=val_transforms)
test_volumes = []
for i in tqdm(range(len(test_ds_fid)), desc="Load test volumes", ncols=100):
    vol_t = test_ds_fid[i]["image"]  # [1,H,W,D] in [-1,1]
    test_volumes.append(vol_t[0].detach().cpu().numpy())

if len(test_volumes) == 0:
    raise RuntimeError("No transformed test volumes available for FID computation.")

# Derive latent shape from a transformed test sample.
sample_t = torch.from_numpy(test_volumes[0]).unsqueeze(0).unsqueeze(0).to(device=device, dtype=torch.float32)
with torch.no_grad():
    latent_shape = autoencoder.encode_stage_2_inputs(sample_t).shape[2:]

fid_metric = FrechetInceptionDistance(feature=2048, normalize=False).to(device)
# Update FID with randomly selected 2D slices from real test volumes.
print("Updating FID with real test images...")
for i in tqdm(range(FID_NUM_REAL), desc="FID real", ncols=100):
    vol = test_volumes[i % len(test_volumes)]
    vol_01 = _to_01_np(vol)
    slice_idx = int(rng.integers(0, vol_01.shape[FID_SLICE_AXIS]))
    sl = _extract_slice(vol_01, FID_SLICE_AXIS, slice_idx)
    fid_img = _slice_to_fid_tensor_uint8(sl, FID_IMAGE_SIZE).to(device)
    fid_metric.update(fid_img, real=True)

print("Generating synthetic samples and updating FID...")
generated_count = 0
scale_for_sampling = float(globals().get("scale_factor_runtime", scale_factor))

# Generate synthetic volumes in small batches to stay within 3D GPU memory limits.
while generated_count < FID_NUM_SYNTH:
    curr_bs = min(FID_GEN_BATCH, FID_NUM_SYNTH - generated_count)

    pred_t1 = generate_t1(
        autoencoder=autoencoder,
        unet=ema_unet,
        scheduler=noise_scheduler,
        scale_factor=scale_for_sampling,
        device=device,
        latent_spatial_shape=latent_shape,
        batch_size=curr_bs,
        latent_channels=latent_channels,
        num_inference_steps=INFERENCE_STEPS,
    )  # [B,1,H,W,D] in [-1,1]

    pred_np = pred_t1[:, 0].detach().cpu().numpy()
    for b in range(curr_bs):
        vol_01 = _to_01_np(pred_np[b])
        slice_idx = int(rng.integers(0, vol_01.shape[FID_SLICE_AXIS]))
        sl = _extract_slice(vol_01, FID_SLICE_AXIS, slice_idx)

        fid_img = _slice_to_fid_tensor_uint8(sl, FID_IMAGE_SIZE).to(device)
        fid_metric.update(fid_img, real=False)

        if SAVE_SYNTHETIC_NIFTI:
            out_path = SYNTH_FID_DIR / f"synth_{generated_count + b:04d}.nii.gz"
            save_nifti_volume(
                pred_t1[b],
                out_path,
                reference_nifti_path=str(test_files[(generated_count + b) % len(test_files)]["image"]),
                match_reference_shape=True,
            )

    generated_count += curr_bs
    if generated_count % 50 == 0 or generated_count == FID_NUM_SYNTH:
        print(f"Generated {generated_count}/{FID_NUM_SYNTH} synthetic samples...")

fid_value = float(fid_metric.compute().detach().cpu().item())
print("=" * 72)
print(f"FID (Generated {FID_NUM_SYNTH} vs Test {FID_NUM_REAL}) = {fid_value:.4f}")
print(f"Test split size (volumes): {len(test_files)}")
print("Note: this is slice-based FID (2D Inception features) computed from 3D volumes.")
print("=" * 72)

# Save metric summary to disk for easy tracking.
fid_report_path = OUTPUT_ROOT / "generated" / "fid_1000_report.txt"
with open(fid_report_path, "w", encoding="utf-8") as f:
    f.write(f"FID_NUM_SYNTH={FID_NUM_SYNTH}\n")
    f.write(f"FID_NUM_REAL={FID_NUM_REAL}\n")
    f.write(f"FID_SLICE_AXIS={FID_SLICE_AXIS}\n")
    f.write(f"FID_IMAGE_SIZE={FID_IMAGE_SIZE}\n")
    f.write(f"FID_VALUE={fid_value:.6f}\n")
    f.write(f"TEST_SPLIT_VOLUMES={len(test_files)}\n")
    f.write("METHOD=slice_based_fid_on_3d_volumes\n")

print(f"Saved FID report: {fid_report_path}")
```
